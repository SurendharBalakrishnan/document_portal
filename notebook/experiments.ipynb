### Embedding Model Check

```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings
```

```python
embedding_model = GoogleGenerativeAIEmbeddings(model="gemini-embedding-2")
```

```python
from langchain_huggingface import HuggingFaceEmbeddings
```

```python
embedding_model = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)
```

```python
embedding_model.embed_query("What is the capital of India?")
```

### Step 1 - Data Ingestion

```python
from langchain_community.document_loaders import PyPDFLoader
```

```python
# pip install langchain-text-splitters
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
import os
```

```python
os.getcwd()  # Get the current working directory
```

```python
file_path = os.path.join(os.getcwd(), "data", "sample.pdf")
```

```python
loader = PyPDFLoader(file_path)
```

```python
documents = loader.load()
```

```python
documents
```

```python
len(documents)  # Get the number of documents loaded
```

### Step 2 Chunking

```python
# This is a experimental value. There is no deterministic way to determine the best chunk size. 
# You can experiment with different values to see what works best for your use case.

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,  # Maximum size of each chunk
    chunk_overlap=150,  # Overlap between chunks
    length_function=len  # Function to calculate the length of text
)
```

```python
docs = text_splitter.split_documents(documents)  # Split the documents into chunks
```

```python
len(docs)  # Get the number of chunks created
```

```python
docs[0]  # Display the first chunk of text
```

```python
docs[0].metadata  # Display the metadata of the first chunk
```

```python
docs[0].page_content  # Display the content of the first chunk
```

### Step 3 Vector Database

```python
# we are using faiss
from langchain_community.vectorstores import FAISS
```

```python
embedding_model.embed_documents(docs[0].page_content)  # Embed the document chunks using the embedding model
```

```python
len(embedding_model.embed_documents(docs[0].page_content))
```

### 1. in memory ( faiss is in memory vector store, chrome)
### 2. on disk storage (faiss you can presist over the disk)
### 3. cloud storage (cloud varient of faiss is not available)(pinecone, weaviate, milvus, mongodbvectorsearch,astradb)

```python
vectorstore = FAISS.from_documents(docs, embedding_model)  # Create a FAISS vector store from the document chunks and embedding model  
```

##### We do similarity search test

### This is Retrieval processess
#### From the vector database we are going to fetch or retrieve or rank the most appropriate k result

```python
relevant_document = vectorstore.similarity_search("llama2 finetuning benchmark experiement.")  # Perform a similarity search for the query "What is the capital of India?" and return the top 2 results
```

```python
relevant_document
```

```python
relevant_document = vectorstore.similarity_search("llama2 finetuning benchmark experiement.", k = 4)
```
