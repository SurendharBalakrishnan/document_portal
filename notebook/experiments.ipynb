### Step 2 Chunking

```python
# This is a experimental value. There is no deterministic way to determine the best chunk size. 
# You can experiment with different values to see what works best for your use case.

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,  # Maximum size of each chunk
    chunk_overlap=200,  # Overlap between chunks
    length_function=len  # Function to calculate the length of text
)
```

```python
docs = text_splitter.split_documents(documents)  # Split the documents into chunks
```

### Step 4 Generation Pipeline

```python
retriever = vectorstore.as_retriever(search_kwargs={"k": 10})  # Convert the vector store to a retriever object for use in retrieval-augmented generation (RAG) tasks  
```

```python
retriever.invoke("llama2 finetuning benchmark experiement.")  # Use the retriever to perform a retrieval-augmented generation (RAG) task with the query "What is the capital of India?"
```

#### Question: user question
#### Context: based on the question retrieving the info from the vector database

```python
prompt_template = """
        Answer the question based on the context provided below. 
        If the context does not contain sufficient information, respond with: 
        "I do not have enough information about this."

        Context: {context}

        Question: {question}

        Answer:"""
```

```python
from langchain_core.prompts import PromptTemplate
```

```python

prompt = PromptTemplate(
    template=prompt_template,
    input_variables=["context", "question"]
)
```

```python
prompt
```

#### create chain template

```python
from langchain_core.output_parsers import StrOutputParser
```

```python
parser = StrOutputParser() # Create an output parser to parse the output of the language model into a string format
```

```python
def format_docs(docs):
    return "\n\n".join([doc.page_content for doc in docs])
```

```python
from langchain_core.runnables import RunnablePassthrough
```

```python
rag_chain = (
    {"context": retriever | format_docs, "question": RunnablePassthrough()}
    | prompt
    | llm
    | StrOutputParser()
)
```

```python
rag_chain.invoke("How was Llama 2-Chat fine-tuned using SFT and RLHF?")  # Use the RAG chain to answer the question "What is the capital of India?" based on the context provided by the retriever 
```

```python
print(rag_chain.invoke("Describe supervised fine-tuning (SFT), reward modeling, rejection sampling and PPO used for Llama 2-Chat"))
```
